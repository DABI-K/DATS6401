```python
import altair as alt
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import streamlit as st

# Dataset: Top 5 EV Makers Sales (Thousands of units)
data = {
    "Year": [2020, 2021, 2022, 2023, 2024],
    "Tesla": [500, 936, 1314, 1808, 1810],
    "BYD": [180, 593, 1863, 3024, 3400],
    "Volkswagen Group": [220, 453, 572, 771, 740],
    "BMW Group": [193, 328, 434, 566, 620],
    "Geely": [68, 100, 329, 672, 850],
}
df = pd.DataFrame(data)
plt.style.use("seaborn-v0_8-whitegrid")
df
```

```python
fig, ax1 = plt.subplots(figsize=(7, 4.5))
for maker in df.columns[1:]:
    ax1.plot(df["Year"], df[maker], marker="o", linewidth=2, label=maker)

ax1.set_title(
    "Chart 1: Annual EV Sales (Position on Common Scale)",
    fontsize=12,
    fontweight="bold",
)
ax1.set_xlabel("Year")
ax1.set_ylabel("Sales (Thousands of Units)")
ax1.set_xticks(df["Year"])
ax1.legend(title="Brand")
plt.tight_layout()
plt.show()
```

```python
fig, ax2 = plt.subplots(figsize=(7, 4.5))
bottom_tracker = np.zeros(len(df["Year"]))

for maker in df.columns[1:]:
    values = df[maker].values
    ax2.bar(
        df["Year"],
        values,
        bottom=bottom_tracker,
        label=maker,
        width=0.55,
        edgecolor="white",
    )
    bottom_tracker += values

ax2.set_title(
    "Chart 2: EV Sales by Brand (Length on Unaligned Scales)",
    fontsize=12,
    fontweight="bold",
)
ax2.set_xlabel("Year")
ax2.set_ylabel("Sales (Thousands of Units)")
ax2.set_xticks(df["Year"])
ax2.legend(title="Brand")
plt.tight_layout()
plt.show()
```

```python
heatmap_matrix = df.set_index("Year").T

fig, ax3 = plt.subplots(figsize=(7, 4))
sns.heatmap(
    heatmap_matrix,
    annot=True,
    fmt="d",
    cmap="YlGnBu",
    cbar_kws={"label": "Sales (Thousands)"},
    ax=ax3,
    linewidths=0.5,
)
ax3.set_title(
    "Chart 3: EV Sales Intensity (Color Saturation Channel)",
    fontsize=12,
    fontweight="bold",
)
ax3.set_xlabel("Year")
ax3.set_ylabel("Brand")
plt.tight_layout()
plt.show()
```
